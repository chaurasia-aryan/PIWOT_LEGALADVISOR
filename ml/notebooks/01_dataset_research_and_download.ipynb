# Notebook 01: Legal Dataset Research, Acquisition & Manifest Verification

This notebook inspects the legal NLP dataset manifest, examines licensing terms (CUAD, LEDGAR, ContractNLI, LexGLUE), and loads the curated legal clause training corpus for the **PIWOT Legal Advisor** ML pipeline.

```python
import os
import json
import pandas as pd

# Paths
DATA_DIR = os.path.abspath(os.path.join("..", "data"))
manifest_csv = os.path.join(DATA_DIR, "dataset_manifest.csv")
manifest_json = os.path.join(DATA_DIR, "dataset_manifest.json")
curated_corpus_path = os.path.join(DATA_DIR, "curated_clause_corpus.json")

print(f"Data Directory: {DATA_DIR}")
```

## 1. Load and Inspect Dataset Manifest

```python
df_manifest = pd.read_csv(manifest_csv)
print(f"Loaded {len(df_manifest)} dataset specifications:\n")
df_manifest[["dataset_id", "dataset_name", "license", "task", "sample_count"]]
```

## 2. Verify Curated Legal Clause Corpus & Category Distribution

```python
with open(curated_corpus_path, "r", encoding="utf-8") as f:
    clauses = json.load(f)

df_clauses = pd.DataFrame(clauses)
print(f"Total Curated Clauses: {len(df_clauses)}")
print("\nClause Count per Category:")
print(df_clauses["category"].value_counts())
print("\nRisk Level Distribution:")
print(df_clauses["risk_level"].value_counts())
df_clauses.head()
```

## 3. Verify Generated Multi-Page Benchmark Contracts

```python
import fitz

sample_dir = os.path.join(DATA_DIR, "sample_contracts")
sample_files = [os.path.join(sample_dir, f) for f in os.listdir(sample_dir) if f.endswith(".pdf")]

for pdf_path in sample_files:
    doc = fitz.open(pdf_path)
    print(f"Contract: {os.path.basename(pdf_path)} | Pages: {len(doc)} | Total Chars: {sum(len(p.get_text()) for p in doc)}")
    doc.close()
```
