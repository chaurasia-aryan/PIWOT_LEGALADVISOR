# Notebook 08: Extractive Contract Summarization (Graph-Based TextRank)

This notebook demonstrates zero-LLM extractive summarization on multi-page contracts using TextRank sentence graph centrality while preserving traceable page citations.

```python
import os
import sys
import json
import pandas as pd

sys.path.insert(0, os.path.abspath(os.path.join("..", "..", "Backend")))

from ml.preprocessing.document_pipeline import process_contract_document
from ml.summarization.extractive_summarizer import ExtractiveContractSummarizer

DATA_DIR = os.path.abspath(os.path.join("..", "data", "sample_contracts"))
msa_pdf = os.path.join(DATA_DIR, "sample_msa_contract.pdf")

with open(msa_pdf, "rb") as f:
    doc_result = process_contract_document(f.read(), filename="sample_msa_contract.pdf")

summarizer = ExtractiveContractSummarizer(max_summary_sentences=4)
summary_res = summarizer.summarize(doc_result["pages"], num_sentences=4)

print(f"Summary Method: {summary_res['method']}")
print(f"Source Pages  : {summary_res['source_pages']}")
print("\n=== EXTRACTED KEY SENTENCES ===")
for i, s in enumerate(summary_res["summary_sentences"], 1):
    print(f"{i}. {s}")
```
